```python
#Enforcing datatypes
df_price['date'] = pd.to_datetime(df_price['date'])
df_news['date'] = pd.to_datetime(df_news['date'])
df_ect['date'] = pd.to_datetime(df_ect['date'])

df_price['ticker'] = df_price['ticker'].astype(str)
df_news['ticker'] = df_news['ticker'].astype(str)
df_ect['ticker'] = df_ect['ticker'].astype(str)

#Ensuring correct panel sorting order across all datasets
df_price = df_price.sort_values(by='date').reset_index(drop=True)
df_news = df_news.sort_values(by='date').reset_index(drop=True)
df_ect = df_ect.sort_values(by='date').reset_index(drop=True)
```

#### News Features + Price

```python
#left_on price means the clean trading-day grid for df_price is preserved

merged_news = pd.merge_asof(
    df_price,
    df_news,
    on='date',
    by='ticker',
    direction='backward', #Matches the most recent historical text sentiment to the current trading day
    tolerance=pd.Timedelta(days=3) #Ensures a cap of max 3 days backwards
)
```

pd.merge_asof perfectly synchronizes irregular, non-synchronous alternative data with a continuous financial timeline without introducing lookahead bias.

```python
merged_news.head(2)
```

#### Transcripts Features + Price 

```python
merged_ects = pd.merge_asof(
    df_price,
    df_ect,
    on='date',
    by='ticker',
    direction='backward',
    tolerance=pd.Timedelta(days=3) 
)
```

```python
merged_ects.head(2)
```

#### Master Combined Dataset (News + Transcripts + Price) 

```python
#News 2019 Filter
news_2019 = merged_news[merged_news['date'].dt.year == 2019].reset_index(drop=True)
```

```python
#ECT 2024 Filter
ect_2024 = merged_ects[merged_ects['date'].dt.year >= 2024].reset_index(drop=True)
```

```python
#News + Price Dataset Filter for 2020 to 2023 overlap period
df_merged_news = merged_news[
    (merged_news['date'].dt.year >= 2020) & 
    (merged_news['date'].dt.year <= 2023)
].reset_index(drop=True)
```

```python
#ECT + Price Dataset Filter for 2020 to 2023 overlap period
df_merged_ect = merged_ects[
    (merged_ects['date'].dt.year >= 2020) & 
    (merged_ects['date'].dt.year <= 2023)
].reset_index(drop=True)
```

```python
#Master Combined Dataset (News + Transcripts + Price) for overlap period (2020-2023)
#Built using the pre-filtered 2020-2023 news dataset to guarantee perfect structural harmony
master_merged = pd.merge_asof(
    df_merged_news, #Filtered news with price for the period 2020-2023
    df_ect, #original ect dataset without price
    on='date', by='ticker', 
    direction='backward', tolerance=pd.Timedelta(days=3)
).reset_index(drop=True)
```

```python
#Lookahead Correlation Check
#Ensures the engineered lag returns and rolling volatilities don't perfectly correlate with targets
max_leakage_corr = 0.0
for ticker, group in master_merged.groupby('ticker'):
    if 'lag_return_1d_winsor' in group.columns and 'movement_T1' in group.columns:
        c = abs(group['lag_return_1d_winsor'].corr(group['movement_T1']))
        if not pd.isna(c) and c > max_leakage_corr:
            max_leakage_corr = c

print(f"Information Coefficient: {max_leakage_corr:.4f}")
assert max_leakage_corr < 0.85, "LEAKAGE DETECTED: Lagged structural components correlate perfectly with future vectors!"
print("Pass: Causal information barrier strictly verified.")
```

```python
#Reviewing NANs
def profile_merge_loss(df, dataset_name, canary_col):
    total_market_days = len(df)
    
    #Matched rows are those where the text feature successfully populated
    matched_rows = df[canary_col].notnull().sum()
    unmatched_rows = df[canary_col].isnull().sum()
    
    match_percentage = (matched_rows / total_market_days) * 100
    loss_percentage = (unmatched_rows / total_market_days) * 100
    
    print(f"Dataset: {dataset_name}")
    print(f"Total Baseline Market Rows: {total_market_days:,}")
    print(f"Matched Observations: {matched_rows:,} ({match_percentage:.2f}%)")
    print(f"Unmatched Gaps (NaNs): {unmatched_rows:,} ({loss_percentage:.2f}%)")
    return unmatched_rows
```

```python
#Reviewing the News merged dataset using a sample column
news_loss = profile_merge_loss(
    df=df_merged_news, 
    dataset_name="DAILY NEWS + PRICE (2020-2023)", 
    canary_col="title_svm_net_sentiment_mean"
)
```

```python
#Reviewing the ECT merged dataset using a sample column
ect_loss = profile_merge_loss(
    df=df_merged_ect, 
    dataset_name="ECTs + PRICE (2020-2023)", 
    canary_col="ect_gpt_max"
)
```

Gaps in the ECT dataset will be more considering the data is sparse as earning call transcripts are quarterly events.
